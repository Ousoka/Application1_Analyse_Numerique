# Résolution analytique des équations différentielles ordinaires (EDO) avec SymPy

### Importation des bibliothèques nécessaires

```python

from sympy import symbols, Function, Eq, dsolve, sin, cos, exp
from sympy.abc import x, C
```

## 1. Définir les variables symboliques et la fonction inconnue
- $x$ : variable indépendante
- $y(x)$ : fonction inconnue dépendant de x

```python
y = Function('y')
```

# 2. Exemple : Équation différentielle linéaire du premier ordre
- Considérons l'équation dy/dx + y = 0
$$\frac{dy}{dx} + y = 0$$

```python
edo1 = Eq(y(x).diff(x) + y(x), 0)
```

- Résolution de l'EDO et Affichage de la Solution

```python
solution1 = dsolve(edo1)
print("Solution générale de l'EDO 1:", solution1)
```

# **Méthode utilisée :**
 Cette équation est une équation différentielle linéaire du premier ordre homogène.
 ### Étapes détaillées :
 1. Réécrire l'équation sous la forme :
    $$\frac{dy}{y} = -dx$$
 2. Intégrer les deux côtés :
    $$\int \frac{1}{y} \,dy = \int -1 \,dx$$
    $$\ln|y| = -x + C$$
 3. Exponentier pour éliminer le logarithme :
    $$y = Ce^{-x}$$
    où $ C = e^C $.
 Cette solution représente la solution générale.

# **Méthode utilisée :**
 Cette équation est une équation différentielle linéaire non homogène.
 ### Étapes détaillées :
 1. Calculer le facteur intégrant :
    $$\mu(x) = e^{\int 1 dx} = e^x$$
 2. Multiplier l'équation par $ \mu(x) $ :
    $$e^x \frac{dy}{dx} + e^x y = e^x \sin(x)$$
    Le membre de gauche devient une dérivée exacte :
    $$\frac{d}{dx}(e^x y) = e^x \sin(x)$$
 3. Intégrer les deux côtés :
    $$e^x y = \int e^x \sin(x) \,dx$$
    En utilisant une intégration par parties :
    $$\int e^x \sin(x) \,dx = \frac{e^x}{2}(\sin(x) - \cos(x)) + C$$
 4. Diviser par $ e^x $ :
    $$y = \frac{1}{2}(\sin(x) - \cos(x)) + Ce^{-x}$$
 Cette solution représente la solution générale.

# **Méthode utilisée :**
 On utilise la méthode du facteur intégrant.
 ### Étapes détaillées :
 1. Calculer le facteur intégrant :
    $$\mu(x) = e^{\int 1 dx} = e^x$$
 2. Multiplier l'équation par $ \mu(x) $ :
    $$e^x \frac{dy}{dx} + e^x y = e^x \cos(x)$$
    $$\frac{d}{dx}(e^x y) = e^x \cos(x)$$
 3. Intégrer :
    $$e^x y = \int e^x \cos(x) \,dx$$
    En utilisant une intégration par parties :
    - On pose $ u = e^x $, $ dv = \cos(x)dx $.
    - Alors, $ du = e^x dx $, $ v = \sin(x) $.
    - Par la formule : $ \int u \, dv = uv - \int v \, du $, on obtient :
      $$\int e^x \cos(x)dx = e^x \sin(x) - \int e^x \sin(x)dx$$
    - En répétant l'intégration par parties pour $ \int e^x \sin(x)dx $, on trouve :
      $$\int e^x \cos(x)dx = \frac{e^x}{2}(\sin(x) + \cos(x)) + C$$
 4. Diviser par $ e^x $ :
    $$y = \frac{1}{2}(\sin(x) + \cos(x)) + Ce^{-x}$$
 5. Appliquer la condition initiale $ y(0) = 2 $ :
    $$2 = \frac{1}{2}(\sin(0) + \cos(0)) + C \quad \Rightarrow \quad 2 = \frac{1}{2}(0 + 1) + C$$
    $$C = \frac{3}{2}$$
    $$y = \frac{1}{2}(\sin(x) + \cos(x)) + \frac{3}{2}e^{-x}$$

# **Méthode utilisée :**
 On utilise la méthode du facteur intégrant.
 ### Étapes détaillées :
 1. Résolution de l'équation homogène associée
  - Avant de résoudre l'équation non homogène, nous résolvons l'équation homogène associée :
    $$\frac{d^2y}{dx^2} - y = 0$$
    L'équation caractéristique de cette équation homogène est :
    $$r^2 - 1 = 0 \quad \Rightarrow \quad r = 1 \quad \text{et} \quad r = -1$$
    Ainsi, la solution générale de l'équation homogène est :
    $$y_h = C_1 e^x + C_2 e^{-x}$$
 2. Trouver une solution particulière
  - Nous cherchons une solution particulière de l'équation non homogène. 
    Le second membre est $\sin(x)$, donc nous supposons que la solution particulière $y_p$ est de la forme :
    $$y_p = A \cos(x) + B \sin(x)$$
    En substituant cette forme dans l'équation, nous trouvons les valeurs de $ A $ et $ B $.
 3. Résolution du système pour $ A $ et $ B $
   - En substituant dans l'EDO, nous obtenons un système d'équations pour les coefficients $ A $ et $ B $. 
     Résolvons ce système pour obtenir la solution particulière.
     Nous obtenons $ A = 0 $ et $ B = -\frac{1}{2} $, donc la solution particulière est :
     $$y_p = -\frac{1}{2} \sin(x)$$
 4. Solution générale de l'EDO
    La solution générale de l'EDO est la somme de la solution homogène et de la solution particulière :
    $$y = C_1 e^x + C_2 e^{-x} - \frac{1}{2} \sin(x)$$
 5. Application des conditions aux limites
    Les conditions aux limites sont données par $y(0) = 0 $ et $ y(\pi) = 1 $.
    - Condition 1 : $ y(0) = 0 $
      Substituons $ x = 0 $ dans la solution générale : $y(0) = C_1 e^0 + C_2 e^0 = C_1 + C_2 = 0$
    - Cela nous donne la première relation entre $ C_1 $ et $ C_2 $ :$C_1 + C_2 = 0 \quad \Rightarrow \quad C_2 = -C_1 $
    - Condition 2 : $ y(\pi) = 1 $
       Substituons $ x = \pi $ dans la solution générale : $y(\pi) = C_1 e^\pi + C_2 e^{-\pi} = 1 $
    - En utilisant $C_2 = -C_1 $, nous avons : $C_1 e^\pi - C_1 e^{-\pi} = 1 $
    - Factorisons par $ C_1 $ : $C_1 (e^\pi - e^{-\pi}) = 1 $
    - Cela nous donne : $C_1 = \frac{1}{e^\pi - e^{-\pi}}$
    - Détermination de $ C_2 $ :
         Puisque $ C_2 = -C_1 $, nous avons : $C_2 = -\frac{1}{e^\pi - e^{-\pi}}$
 6. Solution particulière
     En substituant les valeurs de $ C_1 $ et $ C_2 $ dans la solution générale, la solution particulière est :
     $$y(x) = \frac{1}{e^\pi - e^{-\pi}} \left( e^x - e^{-x} \right) - \frac{1}{2} \sin(x)$$

# **Méthode utilisée :**

 ### Étapes détaillées :
 1. Transformation de l'équation
    Nous allons résoudre cette équation en utilisant la méthode de **variation des paramètres**. 
    La première étape consiste à réécrire l'équation sous une forme plus facile à manipuler.
    Nous divisons l'ensemble de l'équation par $ x $ (en supposant $( x \neq 0)$ :
    $$\frac{dy}{dx} + \frac{2}{x} y = \frac{\sin(x)}{x}$$

2. Calcul du facteur intégrant
   L'équation obtenue est une équation linéaire du premier ordre de la forme :
   $$\frac{dy}{dx} + P(x) y = Q(x)$$
   où $ P(x) = \frac{2}{x} $ et $ Q(x) = \frac{\sin(x)}{x}$.
   Le facteur intégrant $ \mu(x) $ est donné par la formule :
   $$\mu(x) = e^{\int P(x) \, dx}$$
   Substituons $ P(x) = \frac{2}{x} $ dans cette expression :
   $$\mu(x) = e^{\int \frac{2}{x} \, dx} = e^{2 \ln(x)} = x^2$$

3. Multiplication par le facteur intégrant
   Maintenant, nous multiplions l'équation initiale par $\mu(x) = x^2 $ pour rendre le membre de gauche exact :
   $$x^2 \frac{dy}{dx} + 2x y = x \sin(x)$$
   Le membre de gauche peut être réécrit comme une dérivée :
   $$\frac{d}{dx}(x^2 y) = x \sin(x)$$

4. Intégration des deux côtés
   Nous intégrons les deux côtés de l'équation par rapport à $x $ :
   $$\int \frac{d}{dx}(x^2 y) \, dx = \int x \sin(x) \, dx$$
   La première intégrale est triviale et donne :
   $$ x^2 y = \int x \sin(x) \, dx$$
   L'intégrale à droite peut être calculée par la méthode d'intégration par parties. 
   - Soit $ u = x $ et $ dv = \sin(x) , dx $, alors $ du = dx $ et $ v = -\cos(x)$.
   - Appliquons l'intégration par parties :
     $$ \int x \sin(x) \, dx = -x \cos(x) + \int \cos(x) \, dx = -x \cos(x) + \sin(x)$$
   Donc, l'équation devient :
   $$ x^2 y = -x \cos(x) + \sin(x) + C$$

5. Résolution pour $ y $
   Enfin, nous isolons $ y $ en divisant par $ x^2 $ :
   $$ y = \frac{-x \cos(x) + \sin(x) + C}{x^2}$$

6. Solution générale
   La solution générale de l'équation différentielle est donc :
   $$y(x) = \frac{-x \cos(x) + \sin(x)}{x^2} + \frac{C}{x^2}$$

# Résolution d'EDO avec les Méthodes de Runge-Kutta

  Les **équations différentielles ordinaires (EDO)** sont des équations qui relient une fonction inconnue à ses dérivées. Pour résoudre numériquement ces équations, on utilise des méthodes d'intégration. Parmi celles-ci, les **méthodes de Runge-Kutta** sont très populaires et permettent de trouver des solutions approximatives à une EDO de la forme : $$\frac{dy}{dt} = f(t, y), \quad y(t_0) = y_0

  Les méthodes de Runge-Kutta sont définies par une série de calculs successifs des "pentes" intermédiaires, ce qui permet d'obtenir une meilleure approximation à chaque itération.

   ## Développement des Méthodes de Runge-Kutta à partir du Développement de Taylor
  Les méthodes de Runge-Kutta, telles que la méthode d'Euler (RK1), RK2, RK3 et RK4, sont des techniques d'approximation des solutions d'une équation différentielle ordinaire (EDO). Elles sont basées sur un développement en série de Taylor qui permet d'approximer la solution à chaque étape. L'idée centrale est de corriger l'approximation obtenue par les termes d'ordres supérieurs dans la série.

  ### Rappel : Développement de Taylor
 
  Soit une équation différentielle de la forme : $$ \frac{dy}{dt} = f(t, y), \quad y(t_0) = y_0 $$

  Le développement en série de Taylor de la solution $y(t)$ autour de $t_n$ est donné par :
  $$y(t_{n+1}) = y(t_n) + h \cdot f(t_n, y_n) + \frac{h^2}{2!} \cdot f'(t_n, y_n) + \frac{h^3}{3!} \cdot f''(t_n, y_n) + \dots$$

  Ici, $h$ est le pas de temps. La série contient des termes successifs qui sont des dérivées successives de $f(t, y)$. Ces dérivées permettent de calculer les valeurs successives de la fonction solution en fonction du pas de temps et de la solution actuelle.

---

 ## Méthode d'Euler (RK1) : Première approximation

La méthode d'Euler est la plus simple des méthodes de Runge-Kutta et consiste à n'utiliser que le premier terme du développement de Taylor. L'idée est de considérer uniquement la dérivée de la fonction à l'instant $t_n$ pour calculer l'approximation de la solution à $t_{n+1}$.

  ### Développement de Taylor pour la méthode d'Euler

  Le développement de Taylor à l'ordre 1 donne : $$y(t_{n+1}) = y(t_n) + h \cdot f(t_n, y_n) + O(h^2)$$
  Dans cette méthode, l'approximation de la solution à $t_{n+1}$ est simplement : $$y_{n+1} = y_n + h \cdot f(t_n, y_n)$$
  L'erreur associée à cette méthode est d'ordre $O(h^2)$, ce qui signifie qu'elle peut devenir significative pour des pas de temps $h$ relativement grands.

---

 ## Méthode de Runge-Kutta d'ordre 2 (RK2) : Correction à mi-intervalle
 La méthode RK2 améliore l'approximation obtenue par la méthode d'Euler en prenant en compte une correction à mi-chemin de l'intervalle. Elle repose sur l'ajout d'un deuxième terme du développement de Taylor, ce qui permet de mieux capturer les variations de la solution.

### Développement de Taylor pour la méthode RK2
 Le développement de Taylor à l'ordre 2 permet d'introduire la dérivée seconde, et l'approximation de la solution à $t_{n+1}$ devient :
 $$ y(t_{n+1}) = y(t_n) + h \cdot f(t_n, y_n) + \frac{h^2}{2} \cdot f'(t_n, y_n) + O(h^3)$$

 Pour l'implémentation de cette méthode, on choisit une approximation plus précise de la solution à $t_{n+1}$ en calculant deux valeurs successives : une à $t_n$ et l'autre à un point intermédiaire $t_n + \frac{h}{2}$.

 1. Calcul de $k_1 = h \cdot f(t_n, y_n)$,
 2. Calcul de $k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right)$.

 L'approximation de la solution à $t_{n+1}$ est alors donnée par : $$y_{n+1} = y_n + k_2$$

 Cette méthode permet une meilleure approximation que la méthode d'Euler en utilisant une valeur moyenne de la fonction dérivée à mi-intervalle.

---

## Méthode de Runge-Kutta d'ordre 3 (RK3) : Correction par trois dérivées

 La méthode RK3 améliore encore l'approximation en prenant en compte trois termes du développement de Taylor. Elle calcule des valeurs intermédiaires supplémentaires pour mieux suivre l'évolution de la solution.

 ### Développement de Taylor pour la méthode RK3

 Le développement de Taylor à l'ordre 3 permet d'introduire la dérivée troisième, offrant une correction plus précise. Cette méthode calcule trois valeurs successives à différents points de l'intervalle $[t_n, t_{n+1}]$, ce qui améliore la précision.

 Les coefficients $k_1$, $k_2$ et $k_3$ sont calculés comme suit :

 1. **Calcul de $k_1$** : à $t_n$,
    $$
    k_1 = h \cdot f(t_n, y_n)
    $$
 2. **Calcul de $k_2$** : à $t_n + \frac{h}{2}$,
    $$
    k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right)
    $$
 3. **Calcul de $k_3$** : à $t_n + h$,
    $$
    k_3 = h \cdot f\left(t_n + h, y_n - k_1 + 2k_2\right)
    $$

L'approximation de la solution à $t_{n+1}$ est alors une moyenne pondérée des trois termes calculés :$$ y_{n+1} = y_n + \frac{1}{6}(k_1 + 4k_2 + k_3) $$

 Cette méthode prend en compte l'évolution de la solution à trois points différents sur l'intervalle, offrant une meilleure approximation que RK2.

---

## Méthode de Runge-Kutta d'ordre 4 (RK4) : Correction à l'ordre 4
 La méthode RK4 est la méthode la plus couramment utilisée car elle offre une précision d'ordre 4, ce qui permet d'obtenir une solution très précise avec un pas de temps relativement grand.

 ### Développement de Taylor pour la méthode RK4

 Le développement de Taylor à l'ordre 4 tient compte des dérivées jusqu'à la quatrième. Cette méthode utilise quatre termes pour calculer la solution à chaque étape.

 Les coefficients $k_1$, $k_2$, $k_3$ et $k_4$ sont calculés à différents points de l'intervalle :

 1. **Calcul de $k_1$** : à $t_n$,
    $$
    k_1 = h \cdot f(t_n, y_n)
     $$
 2. **Calcul de $k_2$** : à $t_n + \frac{h}{2}$,
    $$
    k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right)
    $$
 3. **Calcul de $k_3$** : à $t_n + \frac{h}{2}$,
    $$
    k_3 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_2}{2}\right)
    $$
 4. **Calcul de $k_4$** : à $t_n + h$,
    $$
    k_4 = h \cdot f(t_n + h, y_n + k_3)
    $$

  L'approximation finale de la solution à $t_{n+1}$ est donnée par une moyenne pondérée des quatre termes :
 
  $$
  y_{n+1} = y_n + \frac{1}{6}(k_1 + 2k_2 + 2k_3 + k_4)
  $$

  Cette méthode est très précise car elle prend en compte l'évolution de la solution à plusieurs points de l'intervalle et utilise des poids pour mieux approximer la solution exacte.

---


  Les méthodes de Runge-Kutta sont basées sur des développements successifs en séries de Taylor, où chaque méthode introduit un nombre croissant de termes pour corriger les erreurs d'approximation. À chaque étape, des coefficients calculés à différents points de l'intervalle sont utilisés pour ajuster l'approximation de la solution de l'EDO :

 - **RK1** : Utilise un seul terme du développement de Taylor.
 - **RK2** : Utilise deux termes pour une meilleure approximation.
 - **RK3** : Introduit un troisième terme pour encore plus de précision.
 - **RK4** : Prend en compte quatre termes pour une approximation très précise.

 Chaque méthode améliore la précision de la solution en augmentant le nombre de corrections apportées au développement de Taylor.

 La méthode de Butcher permet d'organiser systématiquement les coefficients requis pour ces méthodes, facilitant la conception et l'analyse de leurs propriétés.

 | $c_2$    | $a_{21}$ | $a_{22}$ | $\dots$  | $a_{2s}$ |
 | $\vdots$ | $\vdots$ | $\vdots$ | $\ddots$ | $\vdots$ |
 | $c_s$    | $a_{s1}$ | $a_{s2}$ | $\dots$  | $a_{ss}$ |
 |        | $b_1 $   | $b_2$    | $\dots$  | $b_s$    |


 - **$c_i $** : Points intermédiaires d'évaluation.
 - **$a_{ij} $** : Coefficients pour les calculs intermédiaires.
 - **$b_i $** : Poids pour combiner les $k_i $.

## Principe des méthodes de Runge-Kutta
  Les méthodes de Runge-Kutta cherchent à approximer la solution $y(t) $ d'une EDO :$$ \frac{dy}{dt} = f(t, y), \quad y(t_0) = y_0 $$
 1. **Calcul des coefficients $k_i $** : 
 $$ k_i = h \cdot f\left(t_n + c_i h, y_n + \sum_{j=1}^{i-1} a_{ij} k_j\right),  \quad i = 1, \dots, s $$

 2. **Combinaison des coefficients** : $$  y_{n+1} = y_n + \sum_{i=1}^s b_i k_i $$

 ### Conditions générales pour une méthode explicite d'ordre 4

 Pour une méthode de Runge-Kutta explicite d'ordre 4, les coefficients \(a_{ij}\), \(b_i\), et \(c_i\) doivent satisfaire des conditions spécifiques issues du développement en série de Taylor. Ces conditions garantissent la précision de l'ordre 4.

  #### **1. Condition de consistance :**
   La méthode doit être consistante avec la solution de l'équation différentielle :$$\sum_{i=1}^s b_i = 1.$$

  #### **2. Conditions d'ordre 2 :**
   La méthode doit reproduire correctement les termes du développement de Taylor jusqu'à l'ordre 2 : $$\sum_{i=1}^s b_i c_i = \frac{1}{2}.$$  

  #### **3. Conditions d'ordre 3 :**
   Pour un ordre 3, deux relations supplémentaires doivent être satisfaites : $$ \sum_{i=1}^s b_i c_i^2 = \frac{1}{3}, \quad \sum_{i=1}^s \sum_{j=1}^{i-1} b_i b_j a_{ij} c_j = \frac{1}{6}.$$

  #### **4. Conditions d'ordre 4 :**
   Pour garantir un ordre 4, les relations suivantes doivent également être respectées : $$ \sum_{i=1}^s b_i c_i^3 = \frac{1}{4}, \quad \sum_{i=1}^s \sum_{j=1}^{i-1} b_i b_j c_i a_{ij} = \frac{1}{8}.$$

 #### **Propriétés supplémentaires des coefficients**
  - $c_i$ sont les abscisses des points intermédiaires et sont donnés par : $$c_i = \sum_{j=1}^{i-1} a_{ij}.$$
  - $a_{ij}$ est triangulaire inférieur $(a_{ij} = 0$ pour $j \geq i$, ce qui caractérise une méthode explicite.

 Ces conditions sont dérivées en comparant les termes du développement de Taylor de la solution exacte avec ceux de l'approximation fournie par la méthode de Runge-Kutta explicite. Elles déterminent entièrement les coefficients \(a_{ij}\), \(b_i\), et \(c_i\) pour une méthode donnée.
 

## Exemple : RK4
  Le tableau de Butcher pour RK4 :

 | $0$           | $0$           | $0$           | $0$           | $0$           |
 |-------------|-------------|-------------|-------------|-------------|
 | $\frac{1}{2}$ | $\frac{1}{2}$ | $0$           | $0$           | $0$           |
 | $\frac{1}{2}$ | $0$           | $\frac{1}{2}$ | $0 $          | $0 $          |
 | $1$           | $0 $          | $0$           | $1$           | 0$           |
 |             | $\frac{1}{6}$ | $\frac{1}{3}$ | $\frac{1}{3}$ | $\frac{1}{6}$ |


 **Formules de calcul :** $$ \begin{aligned}
k_1 & = h \cdot f(t_n, y_n) \\
k_2 & = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right) \\
k_3 & = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_2}{2}\right) \\
k_4 & = h \cdot f(t_n + h, y_n + k_3)
\end{aligned} $$

$$
y_{n+1} = y_n + \frac{1}{6}(k_1 + 2k_2 + 2k_3 + k_4)
$$

## Exemple pratique
 ### Problème : $\frac{dy}{dt} = -2y $

  **Données :** $f(t, y) = -2y $, $y(0) = 1 $, $h = 0.1 $.

  **Calculs pour $t_1 = t_0 + h = 0.1 $ :**

   1. **$k_1 $** : $  k_1 = 0.1 \cdot (-2 \cdot 1) = -0.2 $

   2. **$k_2 $** : $ k_2 = 0.1 \cdot (-2 \cdot (1 - 0.1)) = -0.18 $

   3. **$k_3 $** : $ k_3 = 0.1 \cdot (-2 \cdot (1 - 0.09)) = -0.182 $

   4. **$k_4 $** : $ k_4 = 0.1 \cdot (-2 \cdot (1 - 0.182)) = -0.1636 $

   5. **Approximation de $y_1 $** :  $$ y_1 = 1 + \frac{1}{6}(-0.2 - 0.36 - 0.364 - 0.1636) = 0.8187 $$
Nous utiliserons un exemple d'EDO simple pour démontrer chacune de ces méthodes.


```python
import numpy as np
import matplotlib.pyplot as plt
```

### Méthode de Runge-Kutta d'ordre 1 : Méthode d'Euler

La méthode d'Euler est la méthode la plus simple parmi les méthodes de Runge-Kutta. Elle repose sur l'approximation suivante :

$$
y_{n+1} = y_n + h \cdot f(t_n, y_n)
$$

Où :
- $h$ est la taille du pas,
- $y_n$ est la valeur approximative de la solution à $t_n$,
- $f(t_n, y_n)$ est la dérivée de la fonction en $t_n$.

Nous allons maintenant implémenter cette méthode pour résoudre une EDO simple.


```python
# Méthode d'Euler (RK1)
def euler(f, t0, y0, t_final, h):
    t_values = np.arange(t0, t_final, h)
    y_values = [y0]
    
    for t in t_values[:-1]:
        y = y_values[-1]
        y_next = y + h * f(t, y)  # Application de la méthode d'Euler
        y_values.append(y_next)
    
    return t_values, np.array(y_values)

# Fonction de l'EDO : dy/dt = y - t^2 + 1
def f(t, y):
    return y - t**2 + 1

# Paramètres
t0 = 0  # Temps initial
y0 = 0.5  # Condition initiale
t_final = 2  # Temps final
h = 0.1  # Pas

# Résolution de l'EDO avec la méthode d'Euler
t_euler, y_euler = euler(f, t0, y0, t_final, h)

# Affichage des résultats
plt.plot(t_euler, y_euler, label='Euler (RK1)', linestyle='-', marker='o')
plt.title('Résolution de l\'EDO avec la méthode d\'Euler (RK1)')
plt.xlabel('Temps t')
plt.ylabel('Solution y(t)')
plt.legend()
plt.grid(True)
plt.show()
```

### Méthode de Runge-Kutta d'ordre 2 : Méthode du Point Milieu

La méthode de Runge-Kutta d'ordre 2 (RK2) est plus précise que la méthode d'Euler. Elle prend en compte la pente au milieu de l'intervalle, ce qui améliore l'approximation de la solution. Voici l'approximation de RK2 :

$$
k_1 = h \cdot f(t_n, y_n)
$$
$$
k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right)
$$
$$
y_{n+1} = y_n + k_2
$$

Implémentons maintenant cette méthode pour résoudre l'EDO.


```python
# Méthode de Runge-Kutta d'ordre 2 (RK2)
def runge_kutta_2(f, t0, y0, t_final, h):
    t_values = np.arange(t0, t_final, h)
    y_values = [y0]
    
    for t in t_values[:-1]:
        y = y_values[-1]
        k1 = h * f(t, y)  # Calcul de k1
        k2 = h * f(t + h/2, y + k1/2)  # Calcul de k2
        y_next = y + k2  # Mise à jour de la solution
        y_values.append(y_next)
    
    return t_values, np.array(y_values)

# Résolution de l'EDO avec la méthode RK2
t_rk2, y_rk2 = runge_kutta_2(f, t0, y0, t_final, h)

# Affichage des résultats
plt.plot(t_rk2, y_rk2, label='Runge-Kutta 2 (RK2)', linestyle='-', marker='x')
plt.title('Résolution de l\'EDO avec la méthode Runge-Kutta 2 (RK2)')
plt.xlabel('Temps t')
plt.ylabel('Solution y(t)')
plt.legend()
plt.grid(True)
plt.show()
```

### Méthode de Runge-Kutta d'ordre 3 : RK3

La méthode RK3 est encore plus précise que RK2. Elle calcule trois pentes intermédiaires pour obtenir une meilleure approximation de la solution :

$$
k_1 = h \cdot f(t_n, y_n)
$$
$$
k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right)
$$
$$
k_3 = h \cdot f\left(t_n + h, y_n - k_1 + 2k_2\right)
$$
$$
y_{n+1} = y_n + \frac{1}{6}(k_1 + 4k_2 + k_3)
$$

Voyons comment cela fonctionne en pratique en implémentant cette méthode.


```python
# Méthode de Runge-Kutta d'ordre 3 (RK3)
def runge_kutta_3(f, t0, y0, t_final, h):
    t_values = np.arange(t0, t_final, h)
    y_values = [y0]
    
    for t in t_values[:-1]:
        y = y_values[-1]
        k1 = h * f(t, y)  # Calcul de k1
        k2 = h * f(t + h/2, y + k1/2)  # Calcul de k2
        k3 = h * f(t + h, y - k1 + 2*k2)  # Calcul de k3
        y_next = y + (k1 + 4*k2 + k3)/6  # Mise à jour de la solution
        y_values.append(y_next)
    
    return t_values, np.array(y_values)

# Résolution de l'EDO avec la méthode RK3
t_rk3, y_rk3 = runge_kutta_3(f, t0, y0, t_final, h)

# Affichage des résultats
plt.plot(t_rk3, y_rk3, label='Runge-Kutta 3 (RK3)', linestyle='-', marker='^')
plt.title('Résolution de l\'EDO avec la méthode Runge-Kutta 3 (RK3)')
plt.xlabel('Temps t')
plt.ylabel('Solution y(t)')
plt.legend()
plt.grid(True)
plt.show()
```

### Méthode de Runge-Kutta d'ordre 4 : RK4

La méthode RK4 est la plus précise parmi les méthodes classiques de Runge-Kutta. Elle utilise quatre pentes intermédiaires pour calculer la solution :

$$
k_1 = h \cdot f(t_n, y_n)
$$
$$
k_2 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_1}{2}\right)
$$
$$
k_3 = h \cdot f\left(t_n + \frac{h}{2}, y_n + \frac{k_2}{2}\right)
$$
$$
k_4 = h \cdot f(t_n + h, y_n + k_3)
$$
$$
y_{n+1} = y_n + \frac{1}{6}(k_1 + 2k_2 + 2k_3 + k_4)
$$

Passons maintenant à l'implémentation de la méthode RK4 pour résoudre l'EDO.


```python
# Méthode de Runge-Kutta d'ordre 4 (RK4)
def runge_kutta_4(f, t0, y0, t_final, h):
    t_values = np.arange(t0, t_final, h)
    y_values = [y0]
    
    for t in t_values[:-1]:
        y = y_values[-1]
        k1 = h * f(t, y)  # Calcul de k1
        k2 = h * f(t + h/2, y + k1/2)  # Calcul de k2
        k3 = h * f(t + h/2, y + k2/2)  # Calcul de k3
        k4 = h * f(t + h, y + k3)  # Calcul de k4
        y_next = y + (k1 + 2*k2 + 2*k3 + k4)/6  # Mise à jour de la solution
        y_values.append(y_next)
    
    return t_values, np.array(y_values)

# Résolution de l'EDO avec la méthode RK4
t_rk4, y_rk4 = runge_kutta_4(f, t0, y0, t_final, h)

# Affichage des résultats
plt.plot(t_rk4, y_rk4, label='Runge-Kutta 4 (RK4)', linestyle='-', marker='s')
plt.title('Résolution de l\'EDO avec la méthode Runge-Kutta 4 (RK4)')
plt.xlabel('Temps t')
plt.ylabel('Solution y(t)')
plt.legend()
plt.grid(True)
plt.show()
```

### Comparaison des Méthodes de Runge-Kutta

Maintenant que nous avons implémenté les méthodes Euler (RK1), Runge-Kutta 2 (RK2), Runge-Kutta 3 (RK3) et Runge-Kutta 4 (RK4), comparons leurs résultats avec la solution exacte de l'EDO.

La solution exacte pour l'EDO $\frac{dy}{dt} = y - t^2 + 1$ est donnée par :

$$
y(t) = (t + 1)^2 - 0.5 \cdot e^t
$$

Nous allons maintenant afficher les résultats des différentes méthodes et la solution exacte sur un même graphique.


```python
# Solution exacte de l'EDO
t_exacte = np.linspace(t0, t_final, 100)
y_exacte = (t_exacte + 1)**2 - 0.5 * np.exp(t_exacte)

# Affichage de la comparaison
plt.plot(t_euler, y_euler, label='Euler (RK1)', linestyle='-', marker='o')
plt.plot(t_rk2, y_rk2, label='Runge-Kutta 2 (RK2)', linestyle='-', marker='x')
plt.plot(t_rk3, y_rk3, label='Runge-Kutta 3 (RK3)', linestyle='-', marker='^')
plt.plot(t_rk4, y_rk4, label='Runge-Kutta 4 (RK4)', linestyle='-', marker='s')
plt.plot(t_exacte, y_exacte, label='Solution exacte', linestyle='--', color='black')

# Personnalisation du graphique
plt.title('Comparaison des méthodes de Runge-Kutta pour la solution d\'une EDO')
plt.xlabel('Temps t')
plt.ylabel('Solution y(t)')
plt.legend()
plt.grid(True)
plt.show()
```
