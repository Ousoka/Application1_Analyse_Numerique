# Problème posé
L'équation est :

$$\frac{\partial u}{\partial t} + \alpha \frac{\partial u}{\partial x} = f \text{ dans } ]0,10] \times [0,T_f]$$

Avec les conditions initiales et limites :

$$u(0,t) = g(t) = x + 1 \text{ sur } [0,T_f]$$

$$u(x,0) = \begin{cases} 
1 &\text{si } x \in ]0,5[ \\
0 &\text{si } x \in [0,10[ 
\end{cases}$$

## Objectifs

```python
import numpy as np
import matplotlib.pyplot as plt

def initial_condition(x):
    return np.where(x < 5, 1, 0)

def boundary_condition(t):
    return t + 1

# Paramètres
a = 1.0
L = 10
Tf = 5
Nx = 100
Nt = 100

dx = L / Nx
dt = Tf / Nt
CFL = a * dt / dx

x = np.linspace(0, L, Nx)
t = np.linspace(0, Tf, Nt)

# Initialisation des solutions
u_left = initial_condition(x)
u_lw = initial_condition(x)

# Schéma décentré gauche
for n in range(1, Nt):
    u_left[1:] = u_left[1:] - CFL * (u_left[1:] - u_left[:-1])
    u_left[0] = boundary_condition(t[n])

# Schéma de Lax-Wendroff
for n in range(1, Nt):
    u_lw[1:-1] = (u_lw[1:-1] - 0.5 * CFL * (u_lw[2:] - u_lw[:-2]) 
                  + 0.5 * CFL**2 * (u_lw[2:] - 2*u_lw[1:-1] + u_lw[:-2]))
    u_lw[0] = boundary_condition(t[n])

# Affichage des résultats
plt.figure(figsize=(10, 6))
plt.plot(x, u_left, label="Décentré gauche")
plt.plot(x, u_lw, label="Lax-Wendroff")
plt.xlabel("x")
plt.ylabel("u(x, t)")
plt.title("Comparaison des schémas")
plt.legend()
plt.grid()
plt.show()
```
