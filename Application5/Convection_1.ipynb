```python
# ===================================================
# Resolution de l Equation d AdveLambdation
# du/dt+ Lambda*du/dx=0 avec x dans[a,b], Lambda>0, et
#   la condition initiale     u(x,0)= x
#   la condition à la limite  u(a,t)= sin(10pi*t)
```

```python
# ==================================================
# IMPORTAION DES LIBRAIRIES
import numpy as np
import matplotlib.pyplot as plt
from numpy import linalg as la
```

```python
#Definition du domaine de Calcul
a = 0; b = 1
# Vitesse d'advection
alpha = 10
```

```python
# MAILLAGE DU DOMAINE

# Nombre de points; points de discrétisation; pas d'espace
NX = 100;x = np.linspace(0.0,1,NX);dx = (b-a)/(NX-1) 

# Choix du Pas de Temps (alpha*dt)/dx = CFL <=1
CFL = 0.5; dt = CFL*dx/alpha; 

# Parametres numerique 
Lambda = (alpha*dt)/dx
ti=0; tf=1
# NT = int(np.floor(1/dt))+1 ; t = 0
NT = int(np.floor((tf-ti)/dt))+1 ; t = 0

#NT = 200

```

```python

# Condition Initiale
def Func1(x):
    return (x)

U0 = np.vectorize(Func1)

# Condition Limite
def Ulim(t):
    return(np.sin(10*np.pi*t))

# Solution Exacte
def Uex(a,b,alpha,x,t,U0,Ulim):
    return (np.where((x-alpha*t)>=a,U0(x-alpha*t),Ulim(t- (x-a)/alpha)))
```

```python
#%% Calcul de la Solution Discrete avec le schema explicite centre

def U_discret1(x,dt,Lambda,NX,NT,U0,Ulim):

    #Initialisation de la Solution Discrete
    U = np.zeros((NX,NT))
    U[:,0] = U0(x)
    t = 0; N = NX-1

    # Boucle en Temps de Calcul de la Solution Discrete
    for n in range(0,NT-1):
        U[1:N,n+1] = U[1:N,n] - (Lambda/2)*( U[2:N+1,n] - U[0:N-1,n] )
        t = t+dt; 
        U[0,n+1] = Ulim(t) # Condition limite
        U[N,n+1] = U[N,n] - (Lambda/2)*( U[N-1,n] - U[N-2,n] )

    return(U)
```

```python
#%% Calcul de la Solution Discrete avec le schema decentre (gauche)

def U_discret2(x,dt,Lambda,NX,NT,U0,Ulim):

    #Initialisation de la Solution Discrete
    U = np.zeros((NX,NT))
    U[:,0] = U0(x)
    t = 0

    # Boucle en Temps de Calcul de la Solution Discrete
    for n in range(0,NT-1):
        U[1:NX,n+1] = U[1:NX,n] - Lambda*( U[1:NX,n] - U[0:NX-1,n] )
        t = t+dt; 
        U[0,n+1] = Ulim(t) # Condition limite

    return(U)
```

```python
#%% Calcul de la Solution Discrete avec le schema de Lax Wendroff

def U_discret3(x,dt,Lambda,NX,NT,U0,Ulim):

    #Initialisation de la Solution Discrete
    U = np.zeros((NX,NT))
    U[:,0] = U0(x)
    t = 0; N = NX-1

    # Boucle en Temps de Calcul de la Solution Discrete
    for n in range(0,NT-1):
        
        U[1:N,n+1] = U[1:N,n] - (Lambda/2)*(U [2:N+1,n] - U[0:N-1,n])  \
        + ((Lambda**2)/2)*(U[2:N+1,n] -2*U[1:N,n] +U[0:N-1,n])
        
        t = t+dt; 

        U[0,n+1] = Ulim(t)

        U[N,n+1] = U[N,n] - (0.5*Lambda)*(U [N-1,n] - U[N-2,n]) \
        + (0.5*(Lambda**2))*(U[N-1,n] -2*U[N,n] +U[N-2,n])

    return(U)   
```

```python
#%% Calcul de la Solution Discrete avec le schema de Lax Frederichs

def U_discret4(x,dt,Lambda,NX,NT,U0,Ulim):   

    #Initialisation de la Solution Discrete
    U = np.zeros((NX,NT))
    U[:,0] = U0(x)
    t = 0; N = NX-1

    # Boucle en Temps de Calcul de la Solution Discrete      
    for n in range(0,NT-1):

         U[1:N,n+1] = (1/2)*(U [2:N+1,n] + U[0:N-1,n])  \
        + ((Lambda)/2)*(U[2:N+1,n]  - U[0:N-1,n])

         t = t+dt; 

         U[0,n+1] = Ulim(t)

         U[N,n+1] = (0.5)*(U [N-1,n] + U[N-2,n]) \
        + (0.5*(Lambda))*(U[N-1,n] - U[N-2,n])

    return(U)    
```

```python
#%% AFFICHAGE DES SOLUTIONS DISCRETES ET EXACTES COMPAREES  

#UU = U_discret1(x,dt,Lambda,NX,NT,U0,Ulim);

UU = U_discret2(x,dt,Lambda,NX,NT,U0,Ulim)

#UU = U_discret3(x,dt,Lambda,NX,NT,U0,Ulim);

# UU = U_discret4(x,dt,Lambda,NX,NT,U0,Ulim);

t  = 0

for n in range(0,NT):

    Uexact = Uex(a,b,alpha,x,t,U0,Ulim)

    if (n%10 == 0): 
        fig, ax = plt.subplots(1) 
        labe = "Solution discrete a t = %1.2f" %(n * dt); 
        ax.plot(x, UU[:,n],'b-o',lw=2, label = labe)
        ax.plot(x, Uexact,'r-+',lw=2, label = 'Solution exacte')
        plt.xlabel('$x$', fontsize=8);plt.ylabel('$U(x,t)$', fontsize=10)
        plt.title('Equation d advection 1D')
        # plt.ylim([-1.1, 1.1])
        # plt.xlim([-0.1, 1.1])
        plt.grid()
        plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.05), shadow=True, ncol=2)
        filename = "ADVECT22/Figure_" + str(int(n/10)) + ".pdf"
        #plt.savefig(filename)

    t = t+dt 
```

```python
#%%============================================================================        

# ETUDE DE LA STABILITE DU SCHEMA

#============================================================================

# Il est question d etudier la condition CFL du Probleme

# La variable CFL = (c*dt)/dx prendra trois valeurs (>1, 1, <1)

# afin de tester la condition CFL<=1 et de voir la stabilite du schema

# pour des pas de temps differents



NX = 100;  x = np.linspace(0.0,1,NX);dx = 1/(NX-1)  
CFLT = np.array([3,1.8,1, 0.5, 0.25]); 
# NT = 500
NC = CFLT.size
dt = CFLT*dx/alpha; NTT = np.zeros(NC)

Lambda = (alpha*dt)/dx

for k in range(0,NC):
    FF = np.floor(1/dt[k])
    NTT = int(FF) + 1
    US = U_discret2(x,dt[k],Lambda[k],NX, NTT,U0,Ulim); t  = 0

    for n in range(0,NTT):
          Uexact = Uex(a,b,alpha,x,t,U0,Ulim)

          if (n%10 == 0): 
              fig, ax = plt.subplots(1); labe = "Solution discrete CFL = %1.2f" %CFLT[k]
              ax.plot(x, US[:,n],'b-o',lw=2, label = labe)
              ax.plot(x, Uexact,'r-',lw=2, label = 'Solution exacte t =%1.2f' %(n * dt[k]))
              plt.xlabel('$x$', fontsize=16);plt.ylabel('$U(x,t)$', fontsize=12)
              plt.title('Schema de Lax-Wendroff')
              plt.ylim([-1.1, 1.1]); plt.xlim([-0.1, 1.1])
              plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.05), shadow=True, ncol=2)
              filename = "ADVECT44/Figure-" +str(CFLT[k]) + "_"+ str(int(n/10))+".pdf"
              # plt.savefig(filename);

          t = t+dt[k]       
```

```python
#%%============================================================================        

# ETUDE DE LA CONSISTENCE, DE LA CONVERGENCE EN ESPACE DU SCHEMA

#============================================================================



NX = np.array([25,50,100,200,400,800,1600]); dx = 1/(NX-1) 
CFLT = 1; 
N = NX.size; NTC = np.zeros(N); dt = (CFLT*dx/alpha)
ERRABS = np.zeros(N); ERRLAT =  ERRABS; 
# ERRABS2 =ERRABS;ERRLAT2 = ERRABS; 

for k in range(0,N):
    IN = int(np.floor(1/dt[k]))+1 
    NTC = int(IN)+1
    x = np.linspace(0.0,1,NX[k])
    Lambda = (alpha*dt[k])/dx[k]

    #UC = U_discret2(x,dt[k],Lambda,NX[k],NTC,U0,Ulim); 
    UC = U_discret3(x,dt[k],Lambda,NX[k],NTC,U0,Ulim)

    t  = 0

    for n in range(0,NTC):

        Uexact = Uex(a,b,alpha,x,t,U0,Ulim)
        t = t+dt[k]

    ERRABS[k] = la.norm(UC[:,n]-Uexact,np.inf); 
    ERRLAT[k] = la.norm(UC[:,n]-Uexact,np.inf) / la.norm(Uexact,np.inf)

    # ERRABS2[k] = la.norm(UC2[:,n]-Uexact,np.inf); 
    # ERRLAT2[k] = la.norm(UC2[:,n]-Uexact,np.inf) / la.norm(Uexact,np.inf) 

    print(); print(ERRLAT[k]," : Erreur Absolue Schema decentre")
    # print(); print(ERRABS2[k]," : Erreur Absolue Schema de L W")

```

```python

#%%

plt.figure(1)
plt.plot(NX,ERRABS, 'b-o',lw=2,label='Erreur Absolue en norme $L^\infty$')
plt.xlabel('$N_x$', fontsize=16)
plt.ylabel('$\Vert U -U_e \Vert_\infty$', fontsize=12)
plt.title("Zone de convergence Schema Decentre CFL=0.5")

# plt.xlim([10, 4200]);plt.ylim([0., 0.01]) 

plt.grid(); plt.legend()

#filename = "ADVECT55/Figure-SD_Zc3"+".pdf"

# plt.savefig(filename)

```
