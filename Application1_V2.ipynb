# Application 1 : Équation de diffusion-réaction généralisée

On considère l’équation de diffusion-réaction stationnaire, où $\mu$, $\gamma$ sont des constantes positives : 

$$
\begin{cases}
-\mu u''(x) + \gamma u(x) = 0 & \text{dans } ]0, 1[ \\
u(0) = 0, \text{ } u(1) = 1
\end{cases} \quad  \text{ } (5)
$$

où : 

- $\mu > 0$ est le coefficient de diffusion,  
- $\gamma > 0$ est le coefficient de réaction.  

## Questions

1. Donner la solution analytique de (5).  

2. Écrire un programme simulant la solution de (5) et la comparer à la solution exacte, pour le même pas d’espace $h$ et pour un choix de $\mu$ et $\gamma$, tel que :  
   - a) $\frac{\gamma}{6\mu} \ll 1$  
   - b) $\frac{\gamma}{6\mu} \gg 1$  

   Ce nombre est le **Péclet**, qui mesure le rapport entre les phénomènes diffusifs et réactifs.  
   - S’il est $\gg 1$, c’est la réaction qui domine sur la diffusion.  
   - S’il est $\ll 1$, c’est la diffusion qui domine.  

3. Pour $\frac{\gamma}{6\mu} \gg 1$ (par exemple de l’ordre de 1000), varier le pas d’espace $h$ en prenant :  
   - $h^2 \frac{\gamma}{6\mu} \leq 1$  
   - $h^2 \frac{\gamma}{6\mu} \geq 1$.  

   Ce nombre est le **Péclet local**.  

4. Quel critère de stabilité pouvez-vous en déduire ?

## 1. Solution analytique de l'équation

L'équation donnée est :

$$
-\mu u''(x) + \gamma u(x) = 0, \quad u(0) = 0 \quad \text{et} \quad u(1) = 1.
$$

**Forme générale de la solution**

L'équation est une EDO linéaire à coefficients constants. 

$$
-\mu \frac{d^2 u(x)}{dx^2} + \gamma \frac{du(x)}{dx} = 0
$$

ou encore

$$
\mu u''(x)  = \gamma u'(x)
$$

et donc 

$$
u''(x)  = \frac{\gamma}{\mu} u'(x)
$$

Posons 

$$
\lambda = \frac{\gamma}{\mu},
$$ 

ce qui donne :

$$
u''(x) - \lambda u'(x) = 0
$$

Cette équation est une équation différentielle linéaire homogène du second ordre à coefficients constants.



Les équations différentielles linéaires à coefficients constants, comme :

$$
u''(x) - \lambda u'(x) = 0,
$$

ont des solutions fondamentales de la forme $e^{rx}$, car elles impliquent des relations polynomiales sur les coefficients de $u(x)$, $u'(x)$, et $u''(x)$.

**Équation caractéristique :**

On suppose une solution de la forme $u(x) = e^{rx}$.  
En substituant $u(x) = e^{rx}$ dans l'équation, on obtient :

$$
r^2 e^{rx} - \lambda r e^{rx} = 0.
$$

Simplifions par $e^{rx} \neq 0$ :

$$
r^2 - \lambda r = 0.
$$

L'équation caractéristique est donc :

$$
r^2 - \lambda r = 0.
$$

**Résolution de l'équation caractéristique :**

L'équation $r^2 - \lambda r = 0$ est une équation quadratique qui se factorise :

$$
(r)(r - \lambda) = 0.
$$

Les racines de cette équation sont :

$$
r_1 = \lambda \quad \text{et} \quad r_2 = 0.
$$

**Solution générale :**

La solution générale de l'équation est une combinaison linéaire des solutions fondamentales $e^{r_1 x}$ et $e^{r_2 x}$ :

$$
u(x) = C_1 e^{\lambda x} + C_2,
$$

où $C_1$ et $C_2$ sont des constantes d'intégration.

**La solution générale de cette EDO est :**

$$
u(x) = C_1 e^{\lambda x} + C_2.
$$

**Conditions aux limites**

- $u(0) = 0$ implique $C_1 + C_2 = 0$, donc $C_2 = -C_1$.
- $u(1) = 1$ implique :  

$$
C_1 \left( e^{\lambda} - e^{-\lambda} \right) = 1.
$$

En isolant $C_1$ :  

$$
C_1 = \frac{1}{e^{\lambda} - e^{-\lambda}}.
$$

En remplaçant $C_1$ et $C_2$ dans l'expression de $u(x)$ :  

$$
u(x) = C_1 e^{\lambda x} + (-C_1) e^{-\lambda x}.
$$

Ce qui devient :  

$$
u(x) = C_1 \left( e^{\lambda x} - e^{-\lambda x} \right).
$$

En remplaçant $C_1$ par son expression :  

$$
u(x) = \frac{1}{e^{\lambda} - e^{-\lambda}} \left( e^{\lambda x} - e^{-\lambda x} \right).
$$

En simplifiant :  

$$
u(x) = \frac{e^{\lambda x} - e^{-\lambda x}}{e^{\lambda} - e^{-\lambda}}.
$$

En remplaçant $ \lambda = \frac{\gamma}{\mu} $ :  

$$
u(x) = \frac{e^{\frac{\gamma x}{\mu}} - e^{-\frac{\gamma x}{\mu}}}{e^{\frac{\gamma}{\mu}} - e^{-\frac{\gamma}{\mu}}}.
$$

**La solution finale devient donc :**  

$$
u(x) = \frac{e^{\frac{\gamma x}{\mu}}}{e^{\frac{\gamma}{\mu}} - 1} - \frac{1}{e^{\frac{\gamma}{\mu}} - 1}.
$$


```python
from sympy import symbols, Function, Eq, dsolve, exp

# Define variables and function
x = symbols('x')
y = Function('y')
mu, gamma = symbols('mu gamma')

# Define the equation
edp = Eq(-mu * y(x).diff(x, 2) + gamma * y(x).diff(x), 0)

# Solve the equation with boundary conditions
solution_analytique = dsolve(edp, ics={y(0): 0, y(1): 1})

# Display the solution
print("Solution particulière de l'EDO avec condition initiale:\n", solution_analytique)
```

## 2. Écrire un programme simulant la solution de (5) et la comparer à la solution exacte, pour le même pas d’espace $h$ et pour un choix de $\mu$ et $\gamma$, tel que :  
   - a) $\frac{\gamma}{6\mu} \ll 1$  
   - b) $\frac{\gamma}{6\mu} \gg 1$  

   Ce nombre est le **Péclet**, qui mesure le rapport entre les phénomènes diffusifs et réactifs.  
   - S’il est $\gg 1$, c’est la réaction qui domine sur la diffusion.  
   - S’il est $\ll 1$, c’est la diffusion qui domine. 


---

### Résolution numérique de l'équation différentielle 

**1. Rappel de l'équation et discrétisation**

L'équation différentielle est :
$$
-\mu u''(x) + \gamma u(x) = 0,
$$
avec :
- $\mu > 0$ : le coefficient de la dérivée seconde,
- $\gamma > 0$ : le coefficient de $u(x)$,
- Les conditions aux limites : $u(0) = 0$ et $u(1) = 1$.

---

**2. Discrétisation du domaine**

Pour résoudre numériquement cette équation, nous divisons le domaine $[0,1]$ en $N$ points avec un pas $h = \frac{1}{N}$. Les points de discrétisation sont donnés par :
$$
x_i = i \cdot h, \quad \text{où } i = 0, 1, \dots, N.
$$

---

**3. Approximation par différences finies**

**3.1. Approximations des dérivées**

Nous utilisons les différences finies pour approximer les dérivées.

**Dérivée seconde ($u''(x)$) :**

$$
u''(x_i) \approx \frac{u_{i-1} - 2u_i + u_{i+1}}{h^2}.
$$

En substituant cette approximation dans l'équation, nous avons :
$$
-\mu \cdot \frac{u_{i-1} - 2u_i + u_{i+1}}{h^2} + \gamma u_i = 0.
$$

**3.2. Réorganisation de l'équation discrète**

Réorganisons l'équation pour $i = 1, 2, \dots, N-1$ :
$$
-\frac{\mu}{h^2} u_{i-1} + \left( \frac{2\mu}{h^2} + \gamma \right) u_i - \frac{\mu}{h^2} u_{i+1} = 0.
$$

On factorise par $\frac{\mu}{h^2}$ :

$$
\frac{\mu}{h^2} \left( -u_{i-1} + \left( 2 + \frac{\gamma h^2}{\mu} \right) u_i - u_{i+1} \right) = 0.
$$

ou:

$$
\frac{\mu}{h^2} \left( -u_{i-1} + 2u_i + \left( \frac{\gamma h^2}{\mu} \right) u_i - u_{i+1} \right) = 0.
$$

ou encore:

$$
\frac{\mu}{h^2} \left( -u_{i-1} + 2u_i - u_{i+1} \right) + \gamma u_i = 0.
$$

Cette équation discrète relie $u_{i-1}$, $u_i$, et $u_{i+1}$. Nous devons résoudre ce système pour les $u_i$ à l'intérieur du domaine.

---

Le problème s'écrit : 

$$
A \mathbf{U} = \mathbf{F}
$$

avec :

Matrice $A$ :

$$
A = \left( \frac{\mu}{h^2} \left(
\begin{matrix}
2 & -1 & 0 & \cdots & 0 \\
-1 & 2 & -1 & \cdots & 0 \\
0 & -1 & 2 & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
0 & \cdots & 0 & -1 & 2
\end{matrix}
\right)
+ 
\left(
\begin{matrix}
\gamma & 0 & 0 & \cdots & 0 \\
0 & \gamma & 0 & \cdots & 0 \\
\vdots & \vdots & \vdots & \ddots & \vdots \\
0 & \cdots & 0 & \gamma & 0 \\
0 & \cdots & 0 & 0 & \gamma
\end{matrix}
\right)
\right)
$$

Vecteur $\mathbf{U}$ :

$$
\mathbf{U} = 
\begin{pmatrix}
u_1 \\
u_2 \\
\vdots \\
u_{N-1}
\end{pmatrix}
$$

Vecteur $\mathbf{F}$ :

$$
\mathbf{F} = 
\begin{pmatrix}
0 \\
0 \\
\vdots \\
\frac{\mu}{h^2}
\end{pmatrix}
$$


---

**4. Formulation matricielle**

Le système discrétisé peut être écrit sous forme matricielle :
$$
\mathbf{A} \mathbf{U} = \mathbf{F},
$$
où :
- $\mathbf{A}$ est une matrice tridiagonale $(N-1) \times (N-1)$,
- $\mathbf{U} = [u_1, u_2, \dots, u_{N-1}]$ est le vecteur des inconnues,
- $\mathbf{F} = [0, 0, \dots, 0, u_N]$ est le vecteur des termes constants, tenant compte des conditions aux limites.

---

**4.1. Construction de la matrice $\mathbf{A}$**

La matrice tridiagonale $\mathbf{A}$ est définie par les coefficients suivants :
$$
\mathbf{A}_{ii} = \frac{2\mu}{h^2} + \gamma, \quad \mathbf{A}_{i,i-1} = \mathbf{A}_{i,i+1} = -\frac{\mu}{h^2}.
$$

---

**4.2. Termes constants $\mathbf{F}$**

Le vecteur $\mathbf{F}$ est nul partout sauf au dernier élément, où il prend en compte la condition $u(1) = 1$ :
$$
\mathbf{F} = [0, 0, \dots, 0, \frac{\mu}{h^2}].
$$

---


```python
# IMPORTATION DES LIBRAIRIES
# ==========================
import numpy as np
from sympy import symbols, Function, Eq, dsolve, exp
from sympy.abc import x, C
from scipy.sparse import dia_matrix
import sympy as sp
import matplotlib.pyplot as plt
import scipy.sparse as sp
from scipy.sparse.linalg import spsolve
from scipy.linalg import norm
```

```python
# MAILLAGE DU DOMAINE
# =============================
def maillage_domaine(NX): 
    """
    Génère le maillage du domaine spatial.
    """
    x = np.linspace(0, 1, NX)  # Positions discrètes sur le domaine [0,1]
    dx = 1 / (NX - 1)  # Pas de discrétisation
    return x, dx
```

```python
# DÉFINITION DES FONCTIONS
# =========================
def func_f(x):
    """
    Définit la fonction f(x). Ici, f(x) = 0 pour tout x.
    """
    return np.zeros_like(x)

def func_c(x): 
    """
    Définit la fonction c(x). Ici, c(x) = γ (constant).
    """
    return gamma * np.ones_like(x)

def solution_exacte(x):
    """
    Calcule la solution analytique exacte de l'équation.
    """
    return (np.exp(gamma * x / mu) - 1) / (np.exp(gamma / mu) - 1)
    
```

```python
# CALCUL DE LA SOLUTION DISCRETE
# ================================
def U_discret(x, NX, dx, f, c):
    """
    Calcule la solution discrétisée à l'aide de la méthode des différences finies.
    """
    h = dx
    U = np.zeros((NX,))  # Initialisation du vecteur solution
    U[0] = 0  # Condition limite u(0) = 0
    U[-1] = 1  # Condition limite u(1) = 1

    F = f(x)
    C = c(x)

    F[0] = 0
    F[-1] = f(x[-1]) + mu / h**2

    F = F[1:NX-1] # Extrait les valeurs intérieures de f
    C = C[1:NX-1] # Extrait les valeurs intérieures de c

    # Assemblage de la matrice du système
    val_diags = [-np.ones(NX - 2), 2 * np.ones(NX - 2), -np.ones(NX - 2)]  # Coefficients diagonaux
    positions = np.array([-1, 0, 1])  # Positions des diagonales
    A1 = dia_matrix((val_diags, positions), shape=(NX - 2, NX - 2))  # Matrice de Laplacien
    A2 = gamma * np.eye(NX - 2)  # Matrice diagonale pour c(x)
    A = (mu / dx**2) * A1 + A2  # Matrice complète

    Uint = spsolve(A, F)  # Résolution du système linéaire
    U[1:NX-1] = Uint  # Insertion des valeurs intérieures

    return U
```

```python
# PARAMÈTRES DU PROBLÈME
# =======================
NX = 10  # Nombre de points dans le maillage
x, dx = maillage_domaine(NX)
```

a) $\frac{\gamma}{6\mu} \ll 1$  

```python
# CAS 1: γ / 6μ ≪ 1 (diffusion dominante)
mu = 1.0  # Diffusivité
gamma = 0.1  # Réaction (γ/6μ = 0.0167)
U_diffusion = U_discret(x, NX, dx, func_f, func_c)
U_exact_diffusion = solution_exacte(x)
```

b) $\frac{\gamma}{6\mu} \gg 1$ 

```python
# CAS 2: γ / 6μ ≫ 1 (réaction dominante)
mu = 1.0  # Diffusivité
gamma = 100.0  # Réaction (γ/6μ = 16.67)
U_reaction = U_discret(x, NX, dx, func_f, func_c)
U_exact_reaction = solution_exacte(x)
U_exact_reaction_peclet_16 = solution_exacte(x)
```

```python
# AFFICHAGE DES SOLUTIONS
# ========================
fig, ax = plt.subplots(2, 1, figsize=(8, 10))

# CAS 1: Diffusion dominante
ax[0].plot(x, U_diffusion, "r-+", lw=2, label="Solution Numérique (Diffusion Dominante)")
ax[0].plot(x, U_exact_diffusion, "b-o", lw=2, label="Solution Analytique (Diffusion Dominante)")
ax[0].set_title("Comparaison des Solutions: Diffusion Dominante (γ/6μ ≪ 1)")
ax[0].set_xlabel("x")
ax[0].legend()
ax[0].grid()

# CAS 2: Réaction dominante
ax[1].plot(x, U_reaction, "r-+", lw=2, label="Solution Numérique (Réaction Dominante)")
ax[1].plot(x, U_exact_reaction, "b-o", lw=2, label="Solution Analytique (Réaction Dominante)")
ax[1].set_title("Comparaison des Solutions: Réaction Dominante (γ/6μ ≫ 1)")
ax[1].set_xlabel("x")
ax[1].legend()
ax[1].grid()

plt.tight_layout()
plt.show()
```

## 4. Quel critère de stabilité pouvez-vous en déduire ?

## Critère de Stabilité

Le critère de stabilité pour ce type d'équation différentielle par la méthode des différences finies peut être lié à la condition de **Péclet**. En effet, la stabilité de la solution numérique dépend fortement du rapport entre les phénomènes diffusifs et réactifs, mesuré par le nombre de Péclet $ \text{Pe} $.

### Pour $ \text{Pe} \gg 1 $ (domination de la réaction sur la diffusion):
Lorsque $ \text{Pe} $ est très grand (par exemple, $ \text{Pe} = 1000 $), cela indique que la réaction domine sur la diffusion. Dans ce cas, le choix du pas d'espace $ h $ est crucial pour la stabilité. Si le pas d'espace est trop grand, la méthode peut devenir instable, car elle ne peut pas bien capturer les variations rapides dues à la réaction.

### Pour $ \text{Pe} \ll 1 $ (domination de la diffusion sur la réaction):
Lorsque $ \text{Pe} $ est petit, cela signifie que la diffusion domine. Dans ce cas, la méthode numérique sera généralement plus stable, car la diffusion agit comme un lisseur, atténuant les effets des grandes variations de la réaction.

### Condition de stabilité:
La condition de stabilité peut être déduite de la relation entre le pas d'espace $ h $ et le nombre de Péclet. En général, on peut exprimer cette condition sous la forme :

$$
\text{Stabilité} \quad \text{si} \quad h^2 \frac{\gamma}{6\mu} \leq 1
$$

Cela signifie que le pas d'espace $ h $ doit être suffisamment petit lorsque le nombre de Péclet est grand pour garantir la stabilité de la solution numérique. Si $ h $ est trop grand par rapport à $ \frac{\gamma}{6\mu} $, la solution numérique peut devenir instable.

En conclusion, pour garantir la stabilité, il est nécessaire de maintenir un pas d'espace suffisamment petit, en particulier lorsque le nombre de Péclet est élevé (c'est-à-dire lorsque la réaction domine). Ce critère peut être utilisé pour ajuster les paramètres de discrétisation en fonction de la nature du problème.


## Analyse de la stabilité de la méthode numérique

Pour analyser la stabilité de la méthode numérique, il faut examiner les critères de stabilité associés à la discrétisation de l'équation. En général, la stabilité dépend du rapport entre les coefficients physiques (tels que $ \gamma $ et $ \mu $) et le pas d'espace $ h $. Voici une démarche pour déterminer la condition de stabilité.

### 1. Critère de stabilité

Pour une équation discrétisée, la stabilité est souvent reliée au nombre de Péclet local défini comme :

$P_{\text{éclet local}} = \frac{h^2 \gamma}{6\mu}$

La solution discrète devient instable si ce nombre dépasse une certaine limite. Typiquement :

- La méthode est stable si $ \frac{h^2 \gamma}{6\mu} \leq 1 $.
- Elle peut devenir instable lorsque $ \frac{h^2 \gamma}{6\mu} > 1 $.

Cela implique que le pas spatial $ h $ doit être suffisamment petit pour respecter cette condition.

### 2. Quand atteint-on l'instabilité ?

L'instabilité survient lorsque :

$ \frac{h^2 \gamma}{6\mu} > 1 $

Cela se traduit par :

$ h > \sqrt{\frac{6\mu}{\gamma}} $

Pour les paramètres $ \mu = 1.0 $ et $ \gamma = 6000.0 $, cela donne :

$ h > \sqrt{\frac{6 \cdot 1.0}{6000.0}} = 0.0316 $

### 3. Nombre de points $ N_X $ pour l'instabilité

Comme $ h = \frac{1}{N_X - 1} $, la condition devient :

$ \frac{1}{N_X - 1} > 0.0316 $

ou encore :

$ N_X - 1 < \frac{1}{0.0316} \approx 31.62 $

Ainsi, l'instabilité survient pour :

$ N_X < 32.62 \quad \Rightarrow \quad N_X \leq 32 $

### 4. Résumé

- **Stabilité** : $ N_X \geq 33 $ (ou $ h \leq 0.0316 $)
- **Instabilité** : $ N_X \leq 32 $ (ou $ h > 0.0316 $)


```python
def check_stability(NX, mu, gamma):
    h = 1 / (NX - 1)
    peclet_local = h**2 * gamma / (6 * mu)
    is_stable = peclet_local <= 1
    return peclet_local, is_stable

# Paramètres
mu = 1.0
gamma = 6000.0

# Vérification pour plusieurs NX
for NX in range(30, 36):  # Explorer autour du seuil
    peclet_local, is_stable = check_stability(NX, mu, gamma)
    status = "Stable" if is_stable else "Instable"
    print(f"NX = {NX}, Péclet local = {peclet_local:.4f}, Statut : {status}")
```
